```python
import hashlib
```

```python
def substituicao_hash(chave, texto):
    # Cria um dicionário para mapear os caracteres da chave para valores substitutos
    substituicoes = {chave[i]: str(i) for i in range(len(chave))}
    
    # Realiza a substituição dos caracteres no texto original
    texto_substituido = ''.join(substituicoes.get(char, char) for char in texto)
    
    # Calcula o hash do texto substituído
    hash_resultante = hashlib.sha256(texto_substituido.encode()).hexdigest()
    
    return hash_resultante
```

```python
def salvar_chave_secreta(chave_secreta):
    # Salva a chave secreta em um arquivo
    with open('chave_secreta.txt', 'w') as arquivo:
        arquivo.write(chave_secreta)
```

```python
def carregar_chave_secreta():
    # Carrega a chave secreta de um arquivo
    with open('chave_secreta.txt', 'r') as arquivo:
        chave_secreta = arquivo.read()
    
    return chave_secreta
```

```python
# Exemplo de uso
chave = input("Digite a chave para a substituição: ")
texto = input("Digite o texto para gerar o hash: ")
```

```python
# Gera o hash usando a técnica de substituição
hash_gerado = substituicao_hash(chave, texto)
print("Hash gerado:", hash_gerado)
```

```python
# Salva a chave secreta em um arquivo
chave_secreta = input("Digite a chave secreta: ")
salvar_chave_secreta(chave_secreta)
print("Chave secreta salva com sucesso!")
```

```python
# Carrega a chave secreta de um arquivo
chave_secreta = carregar_chave_secreta()
print("Chave secreta carregada:", chave_secreta)
```

```python
# Gera o hash usando a chave secreta
hash_secreto = substituicao_hash(chave_secreta, texto)
print("Hash secreto gerado:", hash_secreto)
```
