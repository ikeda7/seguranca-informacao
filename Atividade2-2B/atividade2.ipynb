```python
import random
```

```python
def gcd(a, b):
    # Calcula o maior divisor comum entre dois números
    while b != 0:
        a, b = b, a % b
    return a
```

```python
def mod_inverse(e, phi):
    # Usa o Algoritmo Estendido de Euclides para encontrar o inverso multiplicativo de e (mod phi)
    d, x, y = 0, 1, phi
    while e > 0:
        x, y = y - (y // e) * e, x
        d, e = e, d
    if d == 1:
        return x % phi
    return None
```

```python
def is_prime(num, k=50):
    # Verifica se um número é provavelmente primo usando o teste de Miller-Rabin
    if num <= 1:
        return False
    if num <= 3:
        return True

    # Escrevendo num - 1 como 2^r * d, onde d é ímpar
    d, r = num - 1, 0
    while d % 2 == 0:
        d //= 2
        r += 1

    for _ in range(k):
        a = random.randint(2, num - 2)
        x = pow(a, d, num)
        if x == 1 or x == num - 1:
            continue

        for _ in range(r - 1):
            x = pow(x, 2, num)
            if x == num - 1:
                break
        else:
            return False

    return True
```

```python
def generate_prime(bits):
    # Gera um número primo aleatório com o número de bits especificado
    while True:
        num = random.getrandbits(bits)
        if is_prime(num):
            return num
```

```python
def generate_keypair(bits):
    # Gera um par de chaves RSA (pública e privada) com o número de bits especificado
    p = generate_prime(bits)
    q = generate_prime(bits)

    n = p * q
    phi = (p - 1) * (q - 1)

    # Seleciona um número 'e' tal que 1 < e < phi e gcd(e, phi) = 1
    e = random.randrange(2, phi)
    while gcd(e, phi) != 1:
        e = random.randrange(2, phi)

    # Calcula o inverso multiplicativo 'd' de 'e' mod phi, para obter a chave privada
    d = mod_inverse(e, phi)

    # Retorna os pares de chaves pública e privada
    return ((e, n), (d, n))

if __name__ == "__main__":
    bits = 1024  # Número de bits para gerar os primos p e q (aumentar para maior segurança)
    public_key, private_key = generate_keypair(bits)
    print("Chave pública (e, n):", public_key)
    print("Chave privada (d, n):", private_key)
```
